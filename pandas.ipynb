```python
marks=pd.Series([85,72,91],index=["anu","ann","ben"])
print(marks)
#iloc for int and loc for labels 
print(marks.iloc[2])
print(marks.loc["ben"])
print(marks.iloc[0:2])
print(marks.loc["anu":"ann"])
```

```python
#multiposition
print(marks.iloc[[0,1,2]])
print(marks[["anu","ann"]])
```

```python
#series methods 
print("mean",marks.mean())
print("sum",marks.sum())
print("max",marks.max())
print("min",marks.min())
print("tot_ele",marks.count())
print("each value count\n",marks.value_counts())
print(marks.sort_values(ascending=False))
```

```python
print(marks.loc["ben"])
print(marks.iloc[0])
print("max",marks.max())
print("min",marks.min())
print("avg",marks.mean())
print("score grtr than 75\n",marks[marks>75])
```
